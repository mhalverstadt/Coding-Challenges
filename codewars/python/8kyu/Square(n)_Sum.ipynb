# Square(n) Sum - CodeWars Challenge

## Problem Description
Complete the square sum function so that it squares each number passed into it and then sums the results together.

**Example:**
- For `[1, 2, 2]` it should return `9` (because 1² + 2² + 2² = 1 + 4 + 4 = 9)

This is an 8kyu challenge from CodeWars.

## 1. Define the Square Sum Function

Let's implement the function that takes a list of numbers and returns the sum of their squares.

```python
def square_sum(l: list) -> int:
    """
    Squares each number in the list and returns the sum of the squares.

    Args:
        l (list): A list of numbers

    Returns:
        int: The sum of squares of all numbers in the list
    """
    return sum(num**2 for num in l)

# Display the function
print("Function defined successfully!")
print(f"Function signature: {square_sum.__name__}{square_sum.__annotations__}")
```

## 2. Test the Function with Examples

Let's test our function with the provided example and some additional test cases.

```python
# Test with the provided example
test_case_1 = [1, 2, 2]
result_1 = square_sum(test_case_1)
print(f"square_sum({test_case_1}) = {result_1}")
print(f"Expected: 9 (1² + 2² + 2² = 1 + 4 + 4 = 9)")
print(f"Test passed: {result_1 == 9}\n")

# Additional test cases
test_cases = [
    ([0, 3, 4, 5], 50),  # 0² + 3² + 4² + 5² = 0 + 9 + 16 + 25 = 50
    ([1], 1),            # Single element
    ([2, 2], 8),         # Two identical elements
    ([10], 100),         # Larger single number
]

print("Additional Test Cases:")
for i, (test_input, expected) in enumerate(test_cases, 1):
    result = square_sum(test_input)
    print(f"Test {i}: square_sum({test_input}) = {result}")
    print(f"Expected: {expected}")
    print(f"Test passed: {result == expected}\n")
```

## 3. Explore Different Input Cases

Let's test our function with various edge cases and special inputs.

```python
# Edge cases
print("Edge Cases:")

# Empty list
empty_result = square_sum([])
print(f"Empty list: square_sum([]) = {empty_result}")

# Negative numbers
negative_result = square_sum([-1, -2, 3])
print(f"Negative numbers: square_sum([-1, -2, 3]) = {negative_result}")
print(f"Calculation: (-1)² + (-2)² + 3² = 1 + 4 + 9 = 14")

# Floating point numbers
float_result = square_sum([1.5, 2.5])
print(f"Float numbers: square_sum([1.5, 2.5]) = {float_result}")
print(f"Calculation: 1.5² + 2.5² = 2.25 + 6.25 = 8.5")

# Large numbers
large_result = square_sum([100, 200])
print(f"Large numbers: square_sum([100, 200]) = {large_result}")
print(f"Calculation: 100² + 200² = 10,000 + 40,000 = 50,000")

# Zero mixed with other numbers
zero_mixed = square_sum([0, 0, 5, 0])
print(f"Zeros mixed: square_sum([0, 0, 5, 0]) = {zero_mixed}")
print(f"Calculation: 0² + 0² + 5² + 0² = 0 + 0 + 25 + 0 = 25")
```

## 4. Alternative Implementation Methods

Let's explore different ways to implement the same functionality.

```python
# Method 1: Using a for loop
def square_sum_loop(l: list) -> int:
    """Implementation using a for loop"""
    total = 0
    for num in l:
        total += num ** 2
    return total

# Method 2: Using map() function
def square_sum_map(l: list) -> int:
    """Implementation using map()"""
    return sum(map(lambda x: x**2, l))

# Method 3: Using list comprehension
def square_sum_list_comp(l: list) -> int:
    """Implementation using list comprehension"""
    return sum([x**2 for x in l])

# Method 4: Using NumPy (if available)
try:
    import numpy as np
    def square_sum_numpy(l: list) -> int:
        """Implementation using NumPy"""
        arr = np.array(l)
        return int(np.sum(arr**2))
    numpy_available = True
except ImportError:
    print("NumPy not available - skipping NumPy implementation")
    numpy_available = False

# Test all methods with the same input
test_input = [1, 2, 2]
print(f"Testing with input: {test_input}")
print(f"Original method:     {square_sum(test_input)}")
print(f"For loop method:     {square_sum_loop(test_input)}")
print(f"Map method:          {square_sum_map(test_input)}")
print(f"List comp method:    {square_sum_list_comp(test_input)}")
if numpy_available:
    print(f"NumPy method:        {square_sum_numpy(test_input)}")

# Verify all methods give the same result
methods = [square_sum, square_sum_loop, square_sum_map, square_sum_list_comp]
if numpy_available:
    methods.append(square_sum_numpy)

results = [method(test_input) for method in methods]
print(f"\nAll methods agree: {all(r == results[0] for r in results)}")
```

## 5. Performance Comparison

Let's compare the performance of different implementation methods using timing.

```python
import time
import random

# Create a larger test dataset
large_test = [random.randint(1, 100) for _ in range(10000)]

def time_function(func, data, iterations=1000):
    """Time a function over multiple iterations"""
    start_time = time.time()
    for _ in range(iterations):
        func(data)
    end_time = time.time()
    return (end_time - start_time) / iterations

# Time each method
methods_to_test = [
    ("Generator Expression", square_sum),
    ("For Loop", square_sum_loop),
    ("Map Function", square_sum_map),
    ("List Comprehension", square_sum_list_comp)
]

if numpy_available:
    methods_to_test.append(("NumPy", square_sum_numpy))

print("Performance Comparison (average time per execution):")
print("=" * 50)

results = []
for name, func in methods_to_test:
    avg_time = time_function(func, large_test)
    results.append((name, avg_time))
    print(f"{name:20}: {avg_time:.6f} seconds")

# Find the fastest method
fastest = min(results, key=lambda x: x[1])
print(f"\nFastest method: {fastest[0]} ({fastest[1]:.6f} seconds)")

# Show relative performance
print("\nRelative Performance (compared to fastest):")
for name, time_taken in results:
    relative = time_taken / fastest[1]
    print(f"{name:20}: {relative:.2f}x")
```

## 6. Visualize the Results

Let's create visualizations to better understand how the function works.

```python
import matplotlib.pyplot as plt

# Example data for visualization
example_data = [1, 2, 2, 3, 4]
squares = [x**2 for x in example_data]
sum_of_squares = sum(squares)

# Create subplots
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(15, 5))

# Plot 1: Original numbers
ax1.bar(range(len(example_data)), example_data, color='skyblue', alpha=0.7)
ax1.set_title('Original Numbers')
ax1.set_xlabel('Index')
ax1.set_ylabel('Value')
ax1.set_xticks(range(len(example_data)))
for i, v in enumerate(example_data):
    ax1.text(i, v + 0.1, str(v), ha='center', va='bottom')

# Plot 2: Squared numbers
ax2.bar(range(len(squares)), squares, color='lightcoral', alpha=0.7)
ax2.set_title('Squared Numbers')
ax2.set_xlabel('Index')
ax2.set_ylabel('Value²')
ax2.set_xticks(range(len(squares)))
for i, v in enumerate(squares):
    ax2.text(i, v + 0.5, str(v), ha='center', va='bottom')

# Plot 3: Cumulative sum
cumulative_sum = []
running_total = 0
for square in squares:
    running_total += square
    cumulative_sum.append(running_total)

ax3.plot(range(len(cumulative_sum)), cumulative_sum, marker='o', color='green', linewidth=2, markersize=8)
ax3.set_title('Cumulative Sum of Squares')
ax3.set_xlabel('Index')
ax3.set_ylabel('Cumulative Sum')
ax3.set_xticks(range(len(cumulative_sum)))
ax3.grid(True, alpha=0.3)
for i, v in enumerate(cumulative_sum):
    ax3.text(i, v + 0.5, str(v), ha='center', va='bottom')

plt.tight_layout()
plt.show()

print(f"Input: {example_data}")
print(f"Squares: {squares}")
print(f"Sum of squares: {sum_of_squares}")
print(f"Verification: square_sum({example_data}) = {square_sum(example_data)}")
```
